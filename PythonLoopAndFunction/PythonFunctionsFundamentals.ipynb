```python
def add(a, b) -> int:
    """Return the sum of two numbers."""
    return a + b

def subtract(a, b) -> int:
    """Return the difference of two numbers."""
    return a - b    

def multiply(a, b) -> int:
    """Return the product of two numbers."""
    return a * b

def divide(a, b) -> float:
    """Return the quotient of two numbers."""
    return a / b
```

```python
var_add =add(2, 3)
var_subtract = subtract(5, 2)
var_multiply = multiply(3, 4)
var_divide = divide(10, 2)

print(var_add, var_subtract, var_multiply, var_divide)
```

```python
#Function to check if number is even or odd
def check_even_odd(number: int) -> str:
    """Return 'Even' if the number is even, otherwise return 'Odd'."""
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(5))
print(check_even_odd(6))
```

```python
#Function to find out the max number from the 3 numbers
def check_max(a,b,c):
    """Return the maximum of three numbers."""
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
print( check_max(3, 7, 9) )
print( check_max(10, 7, 9) )
```

```python
#Function to calculate factorial
def factorial_cal(a):
    """Return the factorial of a number."""
    if a == 0 or a == 1:
        return 1
    else:
        result = 1
        for i in range(1, a + 1):
            result *= i
        return result

print(factorial_cal(4))
print(factorial_cal(5))
```

```python
#Create a function that checks whether a number is prime.
def is_prime(n):
    """Return True if the number is prime, otherwise return False."""
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

#Example usage
print(is_prime(7))  # True
print(is_prime(10)) # False
```

```python
#discount calculation function
def calculate_discount(price, discount=10):
    """Return the price after applying the discount."""
    return price - (price * discount / 100)

#Example usage
print(calculate_discount(100))       # 90.0
print(calculate_discount(100, 20))   # 80.0
```

```python
#Create a function that accepts a string and returns the number of vowels.
def count_vowels(input_string):
    """Return the number of vowels in the input string."""
    vowels = "aeiouAEIOU"
    count = 0
    
    for char in input_string:
        if char in vowels:
            count += 1        
    return count

# Example usage:
print(count_vowels('HelloWorld'))
print(count_vowels('Vowels'))  # Output: 3
```

```python
#Create a function that accepts name age course and returns a formatted student profile.
#Use both positional and keyword arguments while calling it.
def student_profile(name, age, course):
    return f"Student Name: {name}\nAge: {age}\nCourse: {course}"

# Calling the function using positional arguments
profile_positional = student_profile("Alice", 20, "Computer Science")

# Calling the function using keyword arguments
profile_keyword = student_profile(name="Bob", age=22, course="Mechanical Engineering")

# Printing the results
print(profile_positional)
print("\n" + profile_keyword)
```

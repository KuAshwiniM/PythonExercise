```python
#to print numbers from 1 to 100 using a for loop.
start_no = int(input("Enter the starting number: "))
end_no = int(input("Enter the ending number: "))
for i in range(start_no, end_no + 1):
    print(i)
```

```python
#Print all even numbers from 1 to 100
for i in range(1,101):
    if i % 2 == 0:
        print(i)
```

```python
#Print all odd numbers from 1 to 100
for i in range(1,101):
    if i % 2 != 0:
        print(i)
```

```python
#multiplication table of a number using for loop.
inp = int(input("Enter the number to find the table: "))
for i in range(1,21):
    print(f"{inp} x {i} = {inp * i}")
```

```python
#Calculate the sum of numbers from 1 to n using a loop.
n = int(input("Enter the value of n: "))
sum = 0
for i in range(1, n + 1):
    sum += i
print(f"The sum of numbers from 1 to {n} is: {sum}")
```

```python
#Calculate the factorial of a number without using any built-in factorial function
n = int(input("Enter the value of n: "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(f"The factorial of {n} is: {factorial}")
```

```python
numbers = [12, 7, 9, 20, 33, 42, 8, 15]
for num in numbers:
    if num % 3 == 0:
        print(num)
```

```python
#Iterate through

student = {
"name": "Rahul",
"age": 22,
"course": "Data Science",
"city": "Bangalore"
}
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
str = input("Enter a string: ")
count = 0
vowels = "aeiouAEIOU"
for char in str:
    if char in vowels:
        count += 1
print(f"The number of vowels in the string is: {count}")        
```

```python
#Reverse a string using a for loop.
input_str = input("Enter a string: ")
reversed_str = ""
for char in input_str:
    print(char)
    reversed_str = char + reversed_str
    print(reversed_str)
print(f"The reversed string is: {reversed_str}")
```

```python
#Find the largest number from a list without using max().
numbers = [3, 7, 2, 9, 1]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print(f"The largest number in the list is: {largest}")
```

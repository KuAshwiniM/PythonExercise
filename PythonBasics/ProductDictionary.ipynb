```python
laptop = {
"brand": "Dell",
"model": "XPS 15",
"price": 120000,
"ram": "16GB",
"storage": "512GB SSD",
"available": True}

#Display the product
laptop
```

```python
#Print brand
print(laptop["brand"])
#Print model
print(laptop["model"])
#Print price
print(laptop["price"])
```

```python
#Change price
laptop["price"] = 110000

#Add processor information
laptop["processor"] = "Intel Core i7"

#Add GPU information
laptop["gpu"] = "NVIDIA GTX 1650"

laptop
```

```python
#Change RAM to "32GB"
laptop["ram"] = "32GB"
#Remove "available"
laptop.pop("available")
print(laptop)

#Print all keys
laptop.keys()

#Print all values
laptop.values()

#Print all items
laptop.items()

```

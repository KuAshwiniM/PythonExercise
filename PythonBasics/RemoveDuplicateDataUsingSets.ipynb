```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

# Without using loops

# Print the original list
print("Original list:", numbers)

# Convert the list into a set
unique_numbers = set(numbers)
print("Unique numbers (set):", unique_numbers)
```

```python
# Convert the set back into a list
unique_numbers_list = list(unique_numbers)

# Print the number of original elements
print("Number of original elements:", len(numbers))

# Print the number of unique elements
print("Number of unique elements:", len(unique_numbers_list))

# Create another example using duplicate student names
student_names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
print("Original student names:", student_names)
unique_student_names = list(set(student_names))
print("Unique student names:", unique_student_names)
```

```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

print(python_students)
print(java_students)
```

```python
#Find students learning either Python or Java
students_learning_either = python_students.union(java_students)
print("Students learning either Python or Java:", students_learning_either)
```

```python
# Find students learning both
students_learning_both = python_students.intersection(java_students)
print("Students learning both Python and Java:", students_learning_both)
```

```python
# Find students learning only Python
students_learning_only_python = python_students.difference(java_students)
print("Students learning only Python:", students_learning_only_python)
```

```python
# Find students learning only Java
students_learning_only_java = java_students.difference(python_students)
print("Students learning only Java:", students_learning_only_java)
```

```python
# Find students belonging to exactly one group
students_in_exactly_one_group = students_learning_either.difference(students_learning_both)
print("Students belonging to exactly one group:", students_in_exactly_one_group)
```

```python
# Add a new student
python_students.add("Vikram")
print("Python students after adding a new student:", python_students)
```

```python
# Remove a student
python_students.remove("Vikram")
print("Python students after removing a student:", python_students)
```

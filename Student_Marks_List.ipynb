```python
#create a list and print its contents
marks = [78, 85, 90, 67, 88, 92, 76]
print(marks)

#print first element of the list
print("First element:", marks[0])

#print last element of the list
print("Last element:", marks[-1])

#Print elements from index 2 to 5
print("Elements from index 2 to 5:", marks[2:6])
```

```python
#Find the number of elements
length_marks = len(marks)
print("length of list is",length_marks)

#Find maximum marks
max_marks = max(marks)
print("maximum marks is",max_marks)

#Find minimum marks
min_marks = min(marks)
print("minimum marks is",min_marks)

#Find total marks
total_marks = sum(marks)
print("total marks is",total_marks)
```

```python
#ascending order
asc_marks = sorted(marks)
print("ascending order of marks is",asc_marks)

#descending order
desc_marks = sorted(marks, reverse=True)
print("descending order of marks is",desc_marks)
```

```python
marks.append(95)
print("marks after adding new marks is",marks)

marks.append([81, 84])
print("marks after adding new list of marks is",marks)
```

```python
marks.remove(67)
print("marks after removing 67 is",marks)

marks.count(90)
print("number of times 90 in list",marks.count(90))
```

```python
print("index of 88 in list is", marks.index(88))
```

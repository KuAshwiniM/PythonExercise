```python
# Create at least 15 different variables containing examples of
int_var = 10
float_var = 3.14
str_var = "Hello"
bool_var = True
list_var = [1, 2, 3]
tuple_var = (1, 2, 3)
set_var = {1, 2, 3}
dict_var = {"key": "value"}
```

```python
# Use type() to identify the data type of each variable.
print(type(int_var))
print(type(float_var))
print(type(str_var))
print(type(bool_var))
print(type(list_var))
print(type(tuple_var))
print(type(set_var))
print(type(dict_var))
```

```python
# Also demonstrate basic type conversion such as
int_from_str = int("100")
print(int_from_str, type(int_from_str))

float_from_str = float("45.67")
print(float_from_str, type(float_from_str))

str_from_int = str(500)
print(str_from_int, type(str_from_int))

bool_from_int = bool(1)
print(bool_from_int, type(bool_from_int))

list_from_tuple = list((1, 2, 3))
print(list_from_tuple, type(list_from_tuple))

tuple_from_list = tuple([1, 2, 3])
print(tuple_from_list, type(tuple_from_list))

set_from_list = set([1, 2, 2, 3])
print(set_from_list, type(set_from_list))
```

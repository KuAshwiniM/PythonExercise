```python
#A list of dictionaries containing product information
products = [
{"name": "Laptop","price": 70000,"brand": "Dell"},
{"name": "Phone","price": 40000,"brand": "Samsung"},
{"name": "Tablet","price": 30000,"brand": "Apple"}]
products
```

```python
#Print all products
print([product["name"] for product in products])
```

```python
#first product from list of dictionaries
print(products[0]["name"])

#Print second product's price
print(products[1]["price"])

#Print third product's brand
print(products[2]["brand"])
```

```python
#Change first product's price
products[0]["price"] = 75000
products

#Add "rating" to the second product
products[1]["rating"] = 4.5

#Add another product manually
products.append({"name": "Smartwatch", "price": 15000, "brand": "Fitbit"})

#Print the final dataset
products
```

Explain the combination of
List, Dictionary, Indexing, Keys, Values
List: An ordered collection of items in square brackets []. Items have a numeric position.
Dictionary: A collection of data stored in curly braces {} as pairs.
Keys: The unique labels inside a dictionary used to find information.
Values: The actual data attached to each key in a dictionary.
Indexing: The method used to pull a specific item out of a collection. Lists use number indexes (like my_list[0]), while dictionaries use key indexes (like my_dict['key']).
